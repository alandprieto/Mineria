## Limpieza de Datos

> `inplace=True` modifica el DataFrame original

### Manejo de Valores Faltantes

```python
df.isnull().sum()
```

```python
df.dropna(subset=['workout_date'], inplace=True)
df
```

```python
x = df['calories_burned'].mean() #.median() #.mode()
df['calories_burned'].fillna(x, inplace=True)
df.loc[[3]]
```

### Filtrado de Filas

```python
for x in df.index:
    if df.loc[x, 'calories_burned'] > 250:
        df.drop(x, inplace=True)
df
```

### Filtrado de Columnas


```python
df.drop(columns='sessions_last_30_days', inplace=True)
df
```

### Filtrado de Duplicados


```python
df.duplicated()
```

```python
df.drop_duplicates(inplace=True)
df
```
